```python
import pandas as pd
df=pd.read_csv("Week 4.2.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Student", y="Hours_Studied")
plt.ylabel("Hours Studied")
plt.title("Who puts in the most effort")
plt.show
```

```python
import matplotlib.pyplot as plt
avg_Hours=df.groupby("Student") ["Hours_Studied"].mean()
avg_Hours.plot(kind="bar")
plt.title("Average of hours studied per student")
plt.ylabel("Average Hours")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Student", y="Grade")
plt.ylabel("Grades")
plt.title("Who performs best")
plt.show
```

```python
import matplotlib.pyplot as plt
avg_Hours=df.groupby("Student") ["Grade"].mean()
avg_Hours.plot(kind="bar")
plt.title("Who performs best")
plt.ylabel("Grade")
plt.show()
```

Highest to the lowest:

```python
import matplotlib.pyplot as plt
avg_Hours=df.groupby("Student") ["Grade"].mean(). sort_values(ascending=False)
avg_Hours.plot(kind="bar")
plt.title("Who performs best")
plt.ylabel("Grade")
plt.show()
```

```python
import matplotlib.pyplot as plt
avg_Hours=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)
avg_Hours.plot(kind="pie")
plt.title("Who performs best")
plt.show()
```

With percentage:

```python
import matplotlib.pyplot as plt
avg_grades=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)
avg_grades.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%')
plt.title("Who performs best")
plt.show()
```

To make it bigger:

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 8))
avg_grades=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)
avg_grades.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%')
plt.title("Who performs best")
plt.show()
```

Does more studying = better grades?

```python
import matplotlib.pyplot as plt
avg_Hours=df.groupby("Hours_Studied") ["Grade"].mean(). sort_values(ascending=False)
avg_Hours.plot(kind="bar")
plt.title("Grades vs Hours")
plt.ylabel("Grade")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30)
plt.show()
```

To increase the transparency, add x and y labels and a title:

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Does more studying equal better grades?")
plt.xlabel("Hours_studied")
plt.ylabel("Grades")
plt.show()
```

How consistent are study patterns and performance:

Candlestick because we can see how the data has spread, minimum, Q's and the maximum. 

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied"]].boxplot()
```

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("How consistent are the study patterns and performance?")
plt.ylabel("Value")
plt.show()
```

```python
hours_min = df["Hours_Studied"].min()
hours_q1 = df["Hours_Studied"].quantile(0.25)
print("Minimum Hours:", hours_min, "Q1:", hours_q1)

plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("How consistent are the study patterns and performance?")
plt.ylabel("Value")
plt.show()
```
