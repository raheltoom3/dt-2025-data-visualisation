Box plot for each athlete to find out the number of years of practice

```python
import pandas as pd
df=pd.read_csv("sports.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice")
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age")
plt.ylabel("Age")
plt.title("Athlete name and age")
plt.show
```

Age vs Years of Practice
plt.scatter (x,y) makes points for each pair of values
s - controls the size of the bubbles
s=df["Years_of_practice"]*30 means if a student has more years of practice their bubble is larger. *30 is a scaling factor to make a bubble big

```python
plt.scatter(df["Age"], df["Years_of_practice"])
plt.show
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
#plt.title["Age vs Years of practice"]
plt.show
```

Average age per sport 

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Average age per sport")
plt.ylabel("Average age")
plt.show()
```
