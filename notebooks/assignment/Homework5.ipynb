```python
import pandas as pd
df=pd.read_csv("Homework5.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_Streams=df.groupby("Artist") ["Streams"].mean()
avg_Streams.plot(kind="bar")
plt.title("Stream count per artist")
plt.ylabel("Streams (M)")
plt.show()
```

Here I found out which artists have the most streams. 

```python
import matplotlib.pyplot as plt
avg_Streams=df.groupby("Artist") ["Streams"].mean(). sort_values(ascending=False)
avg_Streams.plot(kind="bar")
plt.title("Stream count per artist")
plt.ylabel("Streams (M)")
plt.show()
```

I used a descending sorting command to observe the streams better. I noticed that the list is strongly dominated by The Weekend who has streams above 3 billion.

```python
import matplotlib.pyplot as plt
avg_Streams=df.groupby("Artist") ["Streams"].mean()
avg_Streams.plot(kind="pie")
plt.title("Stream count per artist")
plt.show()
```

I used pie chart to visualize streams per artist. 

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 8))
avg_Streams=df.groupby("Artist")["Streams"].mean(). sort_values(ascending=False)
avg_Streams.plot(kind="pie", x="Artist", y="Streams", legend=False, autopct='%1.1f%%')
plt.title("Who is the most listened to artist?")
plt.show()
```

I used a percentage command on pie chart to observe the streams better. The distribution of streams is much more visible using a pie chart with percentages. 

```python
plt.scatter(df["Streams"], df["Release Year"])
plt.title("Does more studying equal better grades?")
plt.xlabel("Streams (M)")
plt.ylabel("Release Year")
plt.show()
```

Here I observed whether the release year of a song affects its popularity. The data shows that the most popular songs are not recent.

```python
streams_min = df["Streams"].min()
streams_q1 = df["Streams"].quantile(0.25)
print("Minimum Streams:", streams_min, "Q1:", streams_q1)

plt.figure(figsize=(8,6))
df[["Streams"]].boxplot()
plt.title("Distribution of song streams")
plt.ylabel("Streams")
plt.show()
```

I use candlestick chart to visualize how the range and distribution of song stream vary. Most songs have lower stream counts while a few songs have very high streams that stand out from the rest.
