```python
#Here I use Python code to work with URLs. I created a URL having the text and generate it to WordCloud. 


import urllib.request 

text = urllib.request.urlopen("https://raw.githubusercontent.com/raheltoom3/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/text.txt").read().decode("utf-8")


WordCloud().generate_from_text(text).to_image()

#Words that appear more times than one are the biggest! So, we definitely say that name Kusti appears in the text the most.
```

```python
#Here I use a function where I stop the words that are not useful. Also, I changed the background color to white. 

WordCloud(stopwords=["ja", "kes", "mis", "kui", "oma", "on"], background_color="white").generate_from_text(text).to_image()

#I removed the stopwords, so the remaining words represent the content, and there are not any grammatical fillers.
#I changed the background to color to white, so the words would be more visible and contrasted. 
```
