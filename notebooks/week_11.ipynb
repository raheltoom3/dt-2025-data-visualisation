```python
sentence = "lumionmaas"
print("Print first 4 letters:", sentence[0:4]) #start: end
print(sentence[:3])

# extract letters from 5-9
print("Print letters 5-9:", sentence[4:9])
# extract letters from 5th to end 
print(sentence[5:])
# extract the last character 
print(sentence[-1])
# extract last four characters 
print(sentence[-4:])
#extract first letter of the substring starting from 5
print(sentence[5:][0])
#sentence[5:] - this will extract a substring starting from 5
#[0] - this will take the first character of that substring

#extract first 5 letters of the substring starting from 5
print(sentence[5:][0:5])

print(sentence[:]) #extract entire string

#extract every second character
print(sentence[::2])

```

```python
sentence = "Alice went to the park and saw a beautiful bird singing"

words = sentence.split(" ") #split the sentence to words by spaces

print(words)

#extract first word
print(words[0])

#extract the first three words
print(words[0:3])

#extract the last three words
print(words[-3:])

#print number of words
print("The number of words: ", len(words))
```
